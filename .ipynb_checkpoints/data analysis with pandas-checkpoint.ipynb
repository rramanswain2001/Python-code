```python
# How do you filter rows in a DataFrame where the salary column is greater than 50000?
import pandas as pd

data = {
    'name': ['Amit', 'Rahul', 'Sneha', 'Priya'],
    'salary': [40000, 60000, 75000, 30000]
}

df = pd.DataFrame(data)

# Filter rows where salary > 50000
filtered_df = df[df['salary'] > 50000]

print(filtered_df)
```

```python
# Write a Pandas statement to filter rows where age > 25 and city == "Delhi".
import pandas as pd

data = {
    'name': ['Amit', 'Rahul', 'Sneha', 'Priya'],
    'age': [22, 30, 28, 24],
    'city': ['Delhi', 'Mumbai', 'Delhi', 'Delhi']
}

df = pd.DataFrame(data)

filtered_df = df[(df['age'] > 25) & (df['city'] == "Delhi")]

print(filtered_df)

```
