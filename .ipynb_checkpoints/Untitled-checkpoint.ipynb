```python
#Find mean sales by each region.
df.groupby('region')['sales'].mean().reset_index()
```

```python
# Find total revenue by product_category.
df.groupby('product_category')['revenue'].sum().reset_index()
```

```python
# Find number of orders per customer.
df.groupby('cust_id')['order_id'].count().reset_index()
```

```python
#Replace all negative profit values with 0.
df.loc[df['profit'] < 0, 'profit'] = 0
print(df.head())
```
